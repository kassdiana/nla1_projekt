# Příklad 4: Modifikovaná mocninná metoda (Posun spektra)

## Zadání
Upravte implementaci mocninné metody tak, aby fungovala pro nalezení nejmenšího vlastního čísla $\lambda_{1}$ SPD matice $A$. Použijte myšlenku o posunu spektra.

**Postup:**
1. Nalezení dominantního vlastního čísla $\lambda_{n}$ matice $A$.
2. Sestavení matice $B = A - \lambda_{n}I$.
3. Nalezení dominantního vlastního čísla $\sigma_{n}$ matice $B$.
4. Výpočet nejmenšího vlastního čísla $A$ jako $\lambda_{1} = \sigma_{n} + \lambda_{n}$.

```python
import numpy as np

def power_method(A, x0, max_it=100, tol=1e-10, use_residual_stop=True):
    """
    Klasická mocninná metoda pro hledání největšího vlastního čísla v absolutní hodnotě.
    """
    # vypočtěte počáteční hodnotu q:
    q = x0 / np.linalg.norm(x0)

    for k in range(1, max_it + 1):
        # vypočtěte x:
        x = A @ q
        # vypočtěte novou hodnotu q:
        q = x / np.linalg.norm(x)
        # vypočtěte hodnotu lambda:
        lambda_k = q.T @ A @ q

        if use_residual_stop:
            residual = np.linalg.norm(A @ q - lambda_k * q)
            if residual < tol:
                break

    return lambda_k, q, k

def power_method_smallest(A, x0, max_it=100, tol=1e-10):
    """
    Modifikovaná mocninná metoda pro nalezení nejmenšího vlastního čísla SPD matice
    s využitím posunu spektra.
    """
    n = A.shape[0]
    
    # Nalezení největšího vlastního čísla matice A
    lambda_n, _, _ = power_method(A, x0, max_it, tol)
    
    # Sestavení posunuté matice B = A - lambda_n * I
    B = A - lambda_n * np.eye(n)
    
    # Nalezení dominantního vlastního čísla matice B (sigma_n)
    sigma_n, _, _ = power_method(B, x0, max_it, tol)
    
    # Zpětný výpočet nejmenšího vlastního čísla
    lambda_1 = sigma_n + lambda_n
    
    return lambda_1, lambda_n, sigma_n
```

```python
# Testování funkce
np.random.seed(42)
A_test = np.random.rand(5, 5)
A_test = A_test @ A_test.T  # Vytvoření symetrické a pozitivně definitní matice
x0_test = np.ones(5)        # Počáteční vektor

lambda_1, lambda_n, sigma_n = power_method_smallest(A_test, x0_test)

eigenvalues_exact = np.linalg.eigvalsh(A_test)
exact_min = np.min(eigenvalues_exact)

print(f"Největší vlastní číslo A (lambda_n): {lambda_n:.4f}")
print(f"Dominantní číslo matice B (sigma_n): {sigma_n:.4f}")
print(f"Nalezené nejmenší vlastní číslo (lambda_1): {lambda_1:.6f}")
print(f"Přesné nejmenší vlastní číslo z np.linalg:   {exact_min:.6f}")
```

## Zadání
Zvolte vhodnou matici $A \in \mathbb{R}^{3 \times 3}$, proveďte 3 ruční iterace mocninné metody pro nalezení nejmenšího vlastního čísla a porovnejte výsledek s přesným řešením.

```python
A = np.array([
    [6, 2, 0],
    [2, 6, 0],
    [0, 0, 3]
], dtype=float)

x0 = np.array([1, 1, 1], dtype=float)

lambda_1_approx, lambda_n, sigma_n = power_method_smallest(A, x0, max_it=3)

# Přesné řešení pomocí knihovny
exact_eigenvalues = np.linalg.eigvalsh(A)
exact_min = np.min(exact_eigenvalues)

print(f"Ruční odhad nejmenšího vl. čísla (po 3 iteracích): {lambda_1_approx}")
print(f"Přesné nejmenší vlastní číslo (np.linalg):       {exact_min}")
print(f"Absolutní chyba: {abs(lambda_1_approx - exact_min)}")

print(f"Největší vl. číslo (lambda_n): {lambda_n}")
print(f"Vl. číslo posunuté matice (sigma_n): {sigma_n}")
```
